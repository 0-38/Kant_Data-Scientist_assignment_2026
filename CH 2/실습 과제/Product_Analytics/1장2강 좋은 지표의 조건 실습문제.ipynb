Ravenstack은 **유료 구독을 안정적으로 유지하면서 반복 매출을 늘리는 것**을 핵심 목표로 정했습니다.

다음 세 지표를 직접 계산한 뒤 현재 목표를 기준으로 KPI, 보조 지표, 허무 지표로 분류하세요.

이번 실습에서는 `end_date`가 비어 있고 `is_trial`이 `False`인 구독을 **현재 활성 유료 구독**으로 정의합니다.

#### 요구사항

1. 현재 활성 유료 구독 조건을 불리언 변수 `active_paid_mask`로 만드세요.
2. 현재 활성 유료 구독의 `mrr_amount` 합계를 `current_paid_mrr`로 계산하세요.
3. `subscriptions`의 `churn_flag` 평균으로 `subscription_churn_rate`를 계산하세요.
4. `accounts`의 고유한 `account_id` 개수를 `cumulative_accounts`로 계산하세요.
5. 세 지표를 알아보기 쉬운 형식으로 출력하세요.
6. 현재 비즈니스 목표를 기준으로 다음과 같이 분류하고 이유를 설명하세요.
   - 현재 유료 구독 MRR: KPI
   - 전체 구독 이탈률: 보조 지표
   - 누적 가입 고객사 수: 허무 지표 후보

#### 해석 질문

**Q1.** 현재 유료 구독 MRR을 KPI로 볼 수 있는 이유는 무엇인가요?  
**Q2.** 전체 구독 이탈률은 현재 유료 구독 MRR을 이해하는 데 어떻게 도움을 주나요?  
**Q3.** 누적 가입 고객사 수만으로 현재 서비스가 성장한다고 판단하기 어려운 이유는 무엇인가요?

#### 제출 결과

- 세 지표의 계산 코드와 결과
- KPI, 보조 지표, 허무 지표 분류
- 분류 이유
- Q1~Q3 답변

```python
# 필수 1 코드를 작성하세요.

# 1. 현재 활성 유료 구독 조건
active_paid_mask = (subscriptions["end_date"].isna() & (subscriptions["is_trial"] == False))

# 2. 현재 활성 유료 구독 MRR
current_paid_mrr = subscriptions.loc[active_paid_mask,"mrr_amount"].sum()

# 3. 전체 구독 이탈률
subscription_churn_rate = subscriptions["churn_flag"].mean()

# 4. 누적 가입 고객사 수
cumulative_accounts = accounts["account_id"].nunique()

# 5. 결과 출력
print(f"현재 유료 구독 MRR: {current_paid_mrr:,.0f}")
print(f"전체 구독 이탈률: {subscription_churn_rate:.2%}")
print(f"누적 가입 고객사 수: {cumulative_accounts:,}개")

print('\n[분류지표]')
print('현재 유료 구독 MRR: KPI')
print('전체 구독 이탈률: 보조 지표')
print('누적 가입 고객사 수: 허무 지표 후보')
```

    - 이탈률이 높아지면 유료 구독과 반복 매출이 감소할 가능성이 있다.

- **Q3.** 누적 가입 고객사 수만으로 현재 서비스가 성장한다고 판단하기 어려운 이유는 무엇인가요?

## 필수 2. 요금제별 핵심 지표 비교하기

### 문제 2-1. 어느 요금제를 우선적으로 살펴봐야 하는가?

#### 문제 설명

전체 평균만 확인하면 요금제별 차이를 놓칠 수 있습니다. `Basic`, `Pro`, `Enterprise` 요금제별로 구독 규모, 이탈률, 현재 유료 구독 MRR을 비교하세요.

#### 요구사항

1. `subscriptions`에 `is_active_paid` 컬럼을 만드세요.
   - `end_date`가 비어 있고 `is_trial`이 `False`이면 `True`
2. `plan_tier`별로 다음 값을 집계하여 `plan_metrics`를 만드세요.
   - 전체 구독 수
   - 이탈 구독 수
   - 구독 이탈률
   - 현재 활성 유료 구독 수
3. 현재 활성 유료 구독만 사용해 요금제별 MRR 합계를 계산하고 `active_mrr` 컬럼으로 추가하세요.
4. 이탈률은 백분율로, MRR은 천 단위 구분 기호를 사용하여 출력하세요.
5. 현재 유료 구독 MRR이 가장 큰 요금제와 이탈률이 가장 높은 요금제를 확인하세요.
6. 현재 목표를 고려하여 우선적으로 점검할 요금제 하나를 정하고, 데이터 근거와 확인할 개선 방향을 설명하세요.

#### 해석 질문

**Q1.** 요금제별 이탈률을 비교하는 것이 전체 이탈률만 확인하는 것보다 행동 가능성이 높은 이유는 무엇인가요?  
**Q2.** 현재 유료 구독 MRR이 가장 큰 요금제는 무엇인가요?  
**Q3.** 이탈률이 가장 높은 요금제는 무엇인가요?  
**Q4.** 위 두 결과를 함께 보면 어떤 요금제를 우선적으로 점검할 수 있으며, 그 이유는 무엇인가요?

#### 제출 결과

- `plan_metrics` 집계 코드와 결과
- MRR 및 이탈률 기준 요금제 비교
- 우선 점검 대상과 개선 방향
- Q1~Q4 답변

```python
# 필수 2 코드를 작성하세요.


# 1. 현재 활성 유료 구독 여부
subscriptions["is_active_paid"] = (subscriptions["end_date"].isna() & (subscriptions["is_trial"] == False))


# 2. 요금제별 핵심 지표 집계
plan_metrics = (
    subscriptions
    .groupby("plan_tier")
    .agg(
        total_subscriptions=("plan_tier", "size"),        # 전체 구독 수
        churned_subscriptions=("churn_flag", "sum"),      # 이탈 구독 수
        churn_rate=("churn_flag", "mean"),                # 구독 이탈률
        active_paid_subscriptions=("is_active_paid", "sum") # 현재 활성 유료 구독 수
    )
)


# 3. 현재 활성 유료 구독의 요금제별 MRR
active_mrr = (
    subscriptions.loc[subscriptions["is_active_paid"]]
    .groupby("plan_tier")["mrr_amount"]
    .sum()
)

# plan_metrics에 active_mrr 추가
plan_metrics["active_mrr"] = active_mrr

# 활성 구독이 없는 요금제가 있을 경우 NaN → 0
plan_metrics["active_mrr"] = plan_metrics["active_mrr"].fillna(0)


# 4. 보기 좋은 형식으로 출력
display_metrics = plan_metrics.copy()

display_metrics["churn_rate"] = (display_metrics["churn_rate"].map(lambda x: f"{x:.2%}").round(3))
display_metrics["active_mrr"] = (display_metrics["active_mrr"].round(3))

display(display_metrics.reset_index())
```

- **Q1.** 요금제별 이탈률을 비교하는 것이 전체 이탈률만 확인하는 것보다 행동 가능성이 높은 이유는 무엇인가요?  
    - 전체 이탈률은 서비스 전체의 현상을 보여주고, 요금제별 이탈률은 어느 요금를 먼저 점검해야하는 지를 판단한다.
    - 다음 회의가 열린다면 담당 팀이 개선 대상을 명확하게 설정하여 개선점을 보여준다.

- **Q2.** 현재 유료 구독 MRR이 가장 큰 요금제는 무엇인가요?

> 이 과제는 필수 2에서 수행한 그룹별 지표 집계와 해석을 새로운 컬럼에 동일하게 적용하는 문제입니다.

#### 요구사항

1. `accounts`를 `referral_source`별로 그룹화하세요.
2. 다음 값을 집계하여 `source_metrics`를 만드세요.
   - 전체 고객사 수
   - 이탈 고객사 수
   - 고객사 이탈률
3. 고객사 이탈률이 높은 순서로 정렬하세요.

7. 유입 경로별 이탈률을 좋은 지표의 세 조건으로 평가하세요.
   - 행동 가능성
   - 비교 가능성
   - 이해 용이성

#### 해석 질문

**Q1.** 고객사 이탈률이 가장 높은 유입 경로는 무엇인가요?  
**Q2.** 유입 경로별 이탈률은 마케팅팀의 행동으로 어떻게 연결할 수 있나요?  
**Q3.** 유입 경로별 이탈률은 좋은 지표의 세 조건을 충족하나요?   
**Q4.** KPI·보조 지표·허무 지표는 어떻게 다른가요?    
고객사 이탈을 줄이려는 목적에서 각각의 지표 예시와 선정 이유를 제시하세요. 허무 지표는 어떤 맥락에서 성과 판단에 도움이 되지 않는지도 설명하세요.

#### 제출 결과

- `source_metrics` 집계 코드와 결과
- 우선 점검할 유입 경로
- 개선 방향
- 좋은 지표의 조건 평가
- Q1~Q4 답변

```python
# 과제 1 코드를 작성하세요.

# 1~3. referral_source별 고객사 수, 이탈 수, 이탈률 집계
source_metrics = (
    accounts
    .groupby("referral_source")
    .agg(
        total_accounts=("account_id", "nunique"),
        churned_accounts=("churn_flag", "sum"),
        churn_rate=("churn_flag", "mean")
    )
    .sort_values("churn_rate", ascending=False).reset_index()
)

# 출력용 DataFrame
display_source_metrics = source_metrics.copy()

display_source_metrics["churn_rate"] = display_source_metrics["churn_rate"].map(lambda x: f"{x:.2%}")

display(display_source_metrics)

# 이탈률이 가장 높은 유입경로 출력
highest_churn_source = source_metrics.iloc[0]["referral_source"]
highest_churn_rate = source_metrics.iloc[0]["churn_rate"]
highest_churn_accounts = source_metrics.iloc[0]["total_accounts"]

print(
    f"이탈률이 가장 높은 유입 경로는 {highest_churn_source}이며, "
    f"이탈률은 {highest_churn_rate:.2%}, "
    f"고객사 수는 {highest_churn_accounts:,}개입니다."
)
```

### 과제 1 답변 작성란

- **Q1.** 고객사 이탈률이 가장 높은 유입 경로는 무엇인가요?  

    - 고객사 이탈률이 다장 높은 유입경로는 30.21%로 event 유입경로가 가장 높은 것으로 나타났다.

<br>

- **Q2.** 유입 경로별 이탈률은 마케팅팀의 행동으로 어떻게 연결할 수 있나요?  

    - event 유입 고객의 이탈률이 가장 높기 때문에, 이벤트를 통해 들어온 고객이 실제 서비스 이용이나 장기 고객으로 충분히 전환되고 있는지 점검할 필요가 있다.

<br>

- **Q3.** 유입 경로별 이탈률은 좋은 지표의 세 조건을 충족하나요?   
    - 좋은 KPI 조건 3가지는 목표와의 연결성, 명확한 측정가능성, 행동가능성이다.
    - 유입 경로별 이탈률은 목표로 연결 가능하며, 수치로 명확하게 측정이 가능하고, 실제 마케팅의 행동지표로 연결할 수 있다고 보기 때문에
    - 좋은 지표라고 볼 수 있다.

<br>

- **Q4.** KPI·보조 지표·허무 지표는 어떻게 다른가요?    
고객사 이탈을 줄이려는 목적에서 각각의 지표 예시와 선정 이유를 제시하세요. 허무 지표는 어떤 맥락에서 성과 판단에 도움이 되지 않는지도 설명하세요.

    - KPI는 핵심 목표를 측정하는 지표로 고객 이탈률이 선정되며,
    - 보조 지표는 KPI가 변한 이유를 이해하는 데 도움을 주는 지표로 고객사 유입경로에 따라 이탈률이 달라 질 수기 때문에 보조지표로 볼 수 있다. 
    - 광고 노출 수나 조회 수는 증가해도 실제 이탈 감소와 직접 연결되지 않는다면 허무 지표가 될 수 있다.

## 실습 마무리

아래 질문에 답하세요.

1. 이번 실습에서 해결하려고 한 비즈니스 문제는 무엇인가요?
    - 유료 구독을 안정적으로 유지하면서 반복 매출을 늘리기 위한 핵심 지표를 관리하고
    - 어느 대상을 우선적으로 개선해야하는 지 확인했다.

2. 현재 목표를 직접 보여주는 KPI로 어떤 지표를 선택했나요?
    - 현재 활성화된 유료 구독자들이 만드는 매출을 MRR을 선택했다.

3. KPI의 변화를 이해하기 위해 어떤 보조 지표를 확인했나요?
    - 전체 구독 이탈률과 요금제별 이탈률을 확인했다.
    - 이탈률은 현재 또는 미래의 매출 감소로 이어지기 때문이다.

4. 허무 지표 후보를 핵심 성과로 사용할 때 어떤 문제가 생길 수 있나요?
    - 누적 가입 고객사를 핵심 성과로 사용한다면 현재 구독을 하지 않거나,
    - 활동하지 않더라도 실제 성과르 과대평가 할 수 있다.

5. 어떤 분석 결과를 근거로 개선 대상을 정했나요?
    - 요금제별로 MRR과 이탈률을 분석하여 Enterprose를 우선 점검 대상으로 정했다.
