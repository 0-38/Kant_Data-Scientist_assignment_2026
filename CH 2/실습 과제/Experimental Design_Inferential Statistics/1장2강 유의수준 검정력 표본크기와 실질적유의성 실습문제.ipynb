다음 조건에서 그룹당 표본 크기에 따른 검정력을 비교하고, 목표 검정력 0.8을 확보하기 위해 필요한 표본 수를 계산하세요.

**Q2.** 검정력이 낮으면 실제 차이가 존재할 때 어떤 오류의 위험이 커지나요?

#### 제출 결과

- 표본 크기별 검정력

```python
# 필수 1 코드를 작성하세요.


# 1. `TTestIndPower()` 객체를 생성하세요.
# 2. 그룹당 표본 수가 30명, 100명, 300명일 때의 검정력을 각각 계산하세요.
# 3. 표본 크기별 검정력을 소수점 셋째 자리까지 출력하세요.
# 4. 유의수준 0.05, 목표 검정력 0.8, 효과 크기 0.3일 때 그룹당 필요한 표본 수를 계산하세요.
# 5. 계산된 표본 수는 소수점 이하를 올림하여 정수로 출력하세요.
# 6. 세 표본 크기 중 목표 검정력 0.8을 충족하는 경우를 확인하세요.

import math
from statsmodels.stats.power import TTestIndPower

# 1 ~ 3
analysis = TTestIndPower()

power_30 = analysis.power(effect_size=0.3, nobs1 = 30, alpha=0.05)
power_100 = analysis.power(effect_size=0.3, nobs1=100, alpha= 0.05)
power_300 = analysis.power(effect_size=0.3, nobs1=300, alpha=0.05)

print(f'표본크기_30 검정력 : {power_30:.3f}')
print(f'표본크기_100 검정력 : {power_100:.3f}')
print(f'표본크기_300 검정력 : {power_300:.3f}')

# 4
required_n = analysis.solve_power(effect_size=0.3, alpha=0.05, power=0.8)
required_n = math.ceil(required_n)

print(f'그룹당 필요한 표본 수 : {required_n} 명')
```

### 문제 2-1. 중앙 냉방시설 유무에 따른 판매가격 차이 분석

#### Cohen’s d 공식

두 집단의 평균 차이를 합동 표준편차로 나누어 계산합니다.

$$
d = \frac{\bar{x}_1 - \bar{x}_2}{s_p}
$$

$$
s_p = \sqrt{\frac{(n_1-1)s_1^2 + (n_2-1)s_2^2}{n_1+n_2-2}}
$$

- $\bar{x}_1$, $\bar{x}_2$: 각 집단의 평균
- $s_1$, $s_2$: 각 집단의 표본 표준편차
- $n_1$, $n_2$: 각 집단의 표본 수
- $s_p$: 합동 표준편차

#### 제출 결과

- 집단별 표본 수와 평균
- 독립표본 t검정 결과
- 평균 차이와 Cohen’s d
- 통계적·효과크기·회사 기준 판단
- 최종 의사결정 근거
- Q1~Q5 답변

```python
# 필수 2 코드를 작성하세요.


air_yes = df[df['CentralAir'] == 'Y']['SalePrice']
air_no = df[df['CentralAir'] == 'N']['SalePrice']

print(f"air_yes 표본 수 : {len(air_yes)} | 평균가격 : {air_yes.mean():.3f}")
print(f"air_no 표본 수 : {len(air_no)} | 평균가격 : {air_no.mean():.3f}")

# 독립표본 t-test
t_stats, pvalue = stats.ttest_ind(air_yes, air_no)

print(f'통계량 : {t_stats:.3f}, p-value : {pvalue:.3f}')


# Cohen's d
mean_yes = air_yes.mean()
mean_no = air_no.mean()

std_yes = air_yes.std()
std_no = air_no.std()

## Pooled Standard Deviation
n_yes = len(air_yes)
n_no = len(air_no)

pooled_std = np.sqrt((n_yes - 1) * std_yes ** 2 + (n_no - 1) * std_no ** 2 / (n_yes + n_no - 2))

## Cohen's d
cohens_d = (mean_no - mean_yes) / pooled_std

## 독립표본 t 검정
t_stats, pvalue = stats.ttest_ind(air_yes, air_no, equal_var=True)


print()
print(f"평균 차이 : {mean_no - mean_yes}")
print(f"Cohen's d: {cohens_d:.3f}")
print(f"p-value : {pvalue:.3f}")

if pvalue <= 0.05:
    print('통계적으로 유의함')
else:
    print('통계적으로 유의하지 않음')

if cohens_d >= 0.8:
    print('cohens_d 큰 효과')
else:
    print('cohens_d 미미한 효과')
```

```python
# 과제 코드를 작성하세요.

# 1. 주방 품질이 `Gd`인 집단과 `TA`인 집단의 `SalePrice`를 각각 준비하세요.
# 2. 두 집단의 표본 수와 평균 판매가격을 출력하세요.
# 3. 두 집단의 평균 차이에 대한 양측 독립표본 t검정을 수행하세요.
# 4. Cohen’s d를 계산하세요.
# 5. 평균 차이, p-value, Cohen’s d를 출력하세요.

kitchen_gd = df[df['KitchenQual'] == 'Gd']['SalePrice']
kitchen_ta = df[df['KitchenQual'] == 'TA']['SalePrice']

n_gd = len(kitchen_gd)
n_ta = len(kitchen_ta)

mean_gd = kitchen_gd.mean()
mean_ta = kitchen_ta.mean()

std_gd = kitchen_gd.std()
std_ta = kitchen_ta.std()

print(f'주방품질 Gd 표본 수 : {n_gd} | 평균 판매가격 : {mean_gd:.3f}')
print(f'주방품질 TA 표본 수 : {n_ta} | 평균 판매가격 : {mean_ta:.3f}')


## Pooled Standard Deviation
pooled_std = np.sqrt((n_gd - 1) * std_gd ** 2 + (n_ta - 1) * std_ta ** 2 / (n_gd + n_ta - 2))

## Cohen's d
cohens_d = (mean_ta - mean_gd) / pooled_std


# 독립표본 t-test
t_stats, pvalue = stats.ttest_ind(kitchen_gd, kitchen_ta)

print()
print(f'평균 차이 : {mean_ta - mean_gd:.3f}')
print(f"Cohen's d: {cohens_d:.3f}")
print(f"p-value : {pvalue:.3f}")
print(f't 통계량 : {t_stats:.3f}')

# 유의수준
if pvalue <= 0.05:
    print('통계적으로 유의함')
else:
    print('통계적으로 유의하지 않음')

# cohens_d
if cohens_d >= 0.8:
    print('cohens_d 큰 효과')
else:
    print('cohens_d 미미한 효과')
```
